```python
def senses_by_change(t2, target_column):
    gain_or_loss = defaultdict(list)
    for target, change in t2.groupby(target_column):
        changed_years = change.years.unique()
        year2sense = {}
        seen = set()
        last_year_observed = {}
        change = change.sort_values("years")
        for i, row in change.iterrows():
            year1, year2 = row.years.split(", ")
            senses1, senses2 = row.sense_ids_year_1.split(","), row.sense_ids_year_2.split(",")
            for sense in senses2:
                if (sense not in senses1) and (sense not in seen):
                    seen.add(sense)
                    gain_or_loss[target_column].append(target)
                    gain_or_loss["sense"].append(sense)
                    gain_or_loss["type"].append("first_observed")
                    gain_or_loss["year"].append(year2)
            for sense in senses1:
                if (sense not in senses2):
                    last_year_observed[sense] = year1
                    gain_or_loss[target_column].append(target)
                    gain_or_loss["sense"].append(sense)
                    gain_or_loss["type"].append("last_observed")
                    gain_or_loss["year"].append(year1)
    for v in gain_or_loss.values():
        print(len(v))
    result = pd.DataFrame(gain_or_loss).sort_values("sense")
    return result
```

```python
result = senses_by_change(t2, "full_phrase")
result
```

```python
result.to_csv("t3_ours-long-nice.tsv", sep="\t")
```
