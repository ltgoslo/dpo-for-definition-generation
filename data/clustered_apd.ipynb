```python
import pandas as pd
```

```python
data = pd.read_csv("t2_clustered.tsv", sep="\t")
data.head()
```

```python
data.shape
```

```python
apd = pd.read_csv("apd.tsv", sep="\t")
apd.head()
```

```python
assert apd[apd.apd.isna()].shape[0] == 0 # no nans in APD
```

```python
apd.shape
```

```python
merged_df = pd.merge(data, apd.reset_index(drop=True), on=['target', 'years'])
merged_df.head()
```

```python
merged_df = merged_df.sort_values(by="apd", ascending=False)
```

```python
merged_df.head()
```

```python
merged_df.to_csv("t2_clustered-apd.tsv", sep="\t", index=False)
```

```python
merged_df.apd.describe()
```
